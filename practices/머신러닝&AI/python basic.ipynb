### 파이썬 기초 개념
* 이 프로그램은 시간대에 따라 다른 인사말을 출력하는 함수를 포함
    * 함수 정의
        * greet() 함수는 두 개의 매개변수를 받습니다:
        * name: 인사할 사람의 이름
        * time_of_day: 시간대 (기본값은 "아침")
    * 딕셔너리 사용
        * greetings 딕셔너리는 시간대별 인사말을 저장합니다.
        * 키(key)는 시간대, 값(value)은 해당 시간대의 인사말입니다.
    * 문자열 포맷팅
        * f-string을 사용하여 인사말과 이름을 조합합니다.
        * f"{greetings[time_of_day]}, {name}님!" 형식으로 문자열을 생성합니다.
    * 기본값 매개변수
        * time_of_day 매개변수는 기본값이 "아침"으로 설정되어 있어서,
        * 함수 호출 시 시간대를 지정하지 않으면 자동으로 "아침" 인사말을 사용
    * 함수 호출 예시:
        * greet("홍길동") → "좋은 아침입니다, 홍길동님!"
        * greet("김철수", "점심") → "맛있는 점심 드세요, 김철수님!"
        * greet("이영희", "저녁") → "편안한 저녁 되세요, 이영희님!"

````python
greetings = {
    "아침": "좋은 아침입니다",
    "점심": "맛있는 점심 드세요",
    "저녁": "편안한 저녁 되세요"
}

def greet(name, time_of_day="아침"):
    """이름과 시간대를 입력받아 적절한 인사말을 출력하는 함수"""
    print(f"{greetings[time_of_day]}, {name}님!")

print("--- 함수 호출 결과 ---")
greet("홍길동")
greet("김철수", "점심")
greet("이영희", "저녁")
print("-" * 20)
````

### 컬렉션 타입 - 리스트(List)
* 학생들의 이름과 점수 정보를 리스트로 관리하는 코드 구현
* 다음 기능을 구현하세요:
    * 학생 추가: 이름과 점수를 입력 받아 목록에 추가
    * 학생 삭제: 이름을 입력 받아 해당 학생 정보 삭제
    * 성적 수정: 이름을 입력 받아 해당 학생의 점수 수정
    * 전체 목록 출력: 모든 학생의 이름과 점수 출력
    * 통계 출력: 최고 점수, 최저 점수, 평균 점수 계산 및 출력

````python
# 1. 초기 학생 데이터 정의
students = [
    {'name': '홍길동', 'score': 92},
    {'name': '김철수', 'score': 85},
    {'name': '이영희', 'score': 98}
]

print("--- 초기 학생 목록 ---")
for student in students:
    print(f"이름: {student['name']}, 점수: {student['score']}")
print("-" * 20)


# 2. 학생 추가
print("\n[학생 추가]")
new_student = {'name': '박지성', 'score': 88}
students.append(new_student)
print(f"'{new_student['name']}' 학생 정보 추가 완료")
print(f"현재 학생 목록: {students}\n")


# 3. 학생 삭제
print("[학생 삭제]")
name_to_delete = '김철수'
students = [student for student in students if student['name'] != name_to_delete]
print(f"'{name_to_delete}' 학생 정보 삭제 완료")
print(f"현재 학생 목록: {students}\n")


# 4. 성적 수정
print("[성적 수정]")
name_to_modify = '이영희'
new_score = 100
for student in students:
    if student['name'] == name_to_modify:
        student['score'] = new_score
        print(f"'{name_to_modify}' 학생의 점수를 {new_score}점으로 수정 완료")
        break
print(f"현재 학생 목록: {students}\n")


# 5. 전체 목록 출력
print("--- 전체 목록 출력 ---")
for student in students:
    print(f"이름: {student['name']}, 점수: {student['score']}")
print("-" * 20 + "\n")


# 6. 통계 출력
print("--- 통계 출력 ---")
scores = [student['score'] for student in students]
highest = max(scores)
lowest = min(scores)
average = sum(scores) / len(scores)

print(f"최고 점수: {highest}점")
print(f"최저 점수: {lowest}점")
print(f"평균 점수: {average:.2f}점")
print("-" * 20)
````

### 튜플(Tuple)
* 주어진 데이터셋에서 튜플을 활용하여 다음 분석을 수행하세요
    * 연도별 판매량 계산
    * 제품별 평균 가격 계산
    * 최대 판매 지역 찾기
    * 분기별 매출 분석

````python
sales_data = [
    (2020, 1, "노트북", 1200, 100, "서울"),
    (2020, 1, "스마트폰", 800, 200, "부산"),
    (2020, 2, "노트북", 1200, 150, "서울"),
    (2020, 2, "스마트폰", 800, 250, "대구"),
    (2020, 3, "노트북", 1300, 120, "인천"),
    (2020, 3, "스마트폰", 850, 300, "서울"),
    (2020, 4, "노트북", 1300, 130, "부산"),
    (2020, 4, "스마트폰", 850, 350, "서울"),
    (2021, 1, "노트북", 1400, 110, "대구"),
    (2021, 1, "스마트폰", 900, 220, "서울"),
    (2021, 2, "노트북", 1400, 160, "인천"),
    (2021, 2, "스마트폰", 900, 270, "부산"),
    (2021, 3, "노트북", 1500, 130, "서울"),
    (2021, 3, "스마트폰", 950, 320, "대구"),
    (2021, 4, "노트북", 1500, 140, "부산"),
    (2021, 4, "스마트폰", 950, 370, "서울")
]

# --- 1. 연도별 판매량 계산 ---
print("## 1. 연도별 판매량 계산")
yearly_sales = {}

for data in sales_data:
    year = data[0]
    quantity = data[4]
    
    yearly_sales[year] = yearly_sales.get(year, 0) + quantity

for year, total_quantity in yearly_sales.items():
    print(f"- {year}년 총 판매량: {total_quantity:,}개")
print("-" * 30)


# --- 2. 제품별 평균 가격 계산 ---
print("\n## 2. 제품별 평균 가격 계산")
product_prices = {}
for data in sales_data:
    product = data[2]
    price = data[3]
    
    if product not in product_prices:
        product_prices[product] = []
    product_prices[product].append(price)

for product, prices in product_prices.items():
    avg_price = sum(prices) / len(prices)
    print(f"- {product} 평균 가격: {avg_price:,.0f}원")
print("-" * 30)


# --- 3. 최대 판매 지역 찾기 ---
print("\n## 3. 최대 판매 지역 분석")
regional_sales = {}
for data in sales_data:
    price = data[3]
    quantity = data[4]
    region = data[5]
    
    total_revenue = price * quantity
    regional_sales[region] = regional_sales.get(region, 0) + total_revenue

top_region = max(regional_sales, key=regional_sales.get)
print(f"- 최대 판매 지역: **{top_region}** (총 매출: {regional_sales[top_region]:,}원)")
print("-" * 30)


# --- 4. 분기별 매출 분석 ---
print("\n## 4. 분기별 매출 분석")
quarterly_revenue = {}
for data in sales_data:
    quarter = data[1]
    price = data[3]
    quantity = data[4]
    
    total_revenue = price * quantity
    quarterly_revenue[quarter] = quarterly_revenue.get(quarter, 0) + total_revenue

for quarter, revenue in sorted(quarterly_revenue.items()):
    print(f"- {quarter}분기 총 매출: {revenue:,}원")
print("-" * 30)
````

### 딕셔너리(Dictionary)
* 딕셔너리를 활용하여 간단한 주소록 프로그램 작성
    * 연락처 이름을 키로 하고, 전화번호, 이메일, 주소 등의 정보를 값으로 저장
    * 중첩 딕셔너리 구조를 사용하여 각 연락처마다 여러 정보를 저장
    * 연락처 추가, 삭제, 검색, 수정, 모든 연락처 보기 기능을 구현

````python
address_book = {
    '김ㅇㅇ': {
        '전화번호': '010-1234-5678',
        '이메일': 'fesd@example.com',
        '주소': '서울'
    },
    '이ㅇㅇ': {
        '전화번호': '010-8765-4321',
        '이메일': 'asdf@example.com',
        '주소': '부산'
    },
    '박ㅇㅇ': {
        '전화번호': '010-5555-8888',
        '이메일': 'fwew@example.com',
        '주소': '인천'
    }
}

def add_contact(name, phone, email, address):
    """연락처 추가"""
    if name in address_book:
        print(f"'{name}'님은 이미 등록되어 있습니다.")
        return
    address_book[name] = {
        '전화번호': phone,
        '이메일': email,
        '주소': address
    }
    print(f"'{name}'님의 연락처를 추가했습니다.")

def delete_contact(name):
    """연락처 삭제"""
    if name in address_book:
        del address_book[name]
        print(f"'{name}'님의 연락처를 삭제했습니다.")
    else:
        print(f"'{name}'님의 연락처를 찾을 수 없습니다.")
        
def search_contact(name):
    """연락처 검색"""
    if name in address_book:
        info = address_book[name]
        print(f"\n--- '{name}'님 검색 결과 ---")
        print(f"  - 전화번호: {info['전화번호']}")
        print(f"  - 이메일: {info['이메일']}")
        print(f"  - 주소: {info['주소']}")
        print("-" * 25)
    else:
        print(f"'{name}'님의 연락처를 찾을 수 없습니다.")

def modify_contact(name, part_to_change, new_value):
    """연락처 수정"""
    if name in address_book:
        if part_to_change in address_book[name]:
            address_book[name][part_to_change] = new_value
            print(f"'{name}'님의 '{part_to_change}' 정보를 '{new_value}'(으)로 수정했습니다.")
        else:
            print(f"'{part_to_change}' 정보가 없습니다")
    else:
        print(f"'{name}'님의 연락처를 찾을 수 없습니다.")

def view_all_contacts():
    """모든 연락처 보기"""
    print("\n--- 모든 연락처 보기 ---")
    if not address_book:
        print("주소록이 비어있습니다.")
    else:
        for name, info in address_book.items():
            print(f"이름: {name}")
            print(f"  - 전화번호: {info['전화번호']}")
            print(f"  - 이메일: {info['이메일']}")
            print(f"  - 주소: {info['주소']}")
            print()
    print("-" * 25)

add_contact('최ㅇㅇ', '010-1234-5678', 'dfgdf@example.com', '서울')
add_contact('장ㅇㅇ', '010-8765-4321', 'hrte@example.com', '부산')
add_contact('정ㅇㅇ', '010-5555-8888', 'qeas@example.com', '인천')

search_contact('이ㅇㅇ')

modify_contact('김ㅇㅇ', '주소', '수원')

delete_contact('박ㅇㅇ')

view_all_contacts()
````

### 집합(Set)
* 소셜 네트워크에서 사용자 간의 관계와 추천 시스템을 구현하는 프로그램을 작성
    * 공통 관심사를 갖는 친구 응답
    * 공통 관심사가 없는 친구 응답

```python
{
    "Alice": ["음악", "영화", "독서"],
    "Bob": ["스포츠", "여행", "음악"],
    "Charlie": ["프로그래밍", "게임", "영화"],
    "David": ["요리", "여행", "사진"],
    "Eve": ["프로그래밍", "독서", "음악"],
    "Frank": ["스포츠", "게임", "요리"],
    "Grace": ["영화", "여행", "독서"]
}
```

````python
# 사용자별 관심사 데이터
user_interests = {
    "Alice": ["음악", "영화", "독서"],
    "Bob": ["스포츠", "여행", "음악"],
    "Charlie": ["프로그래밍", "게임", "영화"],
    "David": ["요리", "여행", "사진"],
    "Eve": ["프로그래밍", "독서", "음악"],
    "Frank": ["스포츠", "게임", "요리"],
    "Grace": ["영화", "여행", "독서"]
}

def find_friends_with_common_interests(target_user, all_users):
    """공통 관심사를 가진 친구와 그 관심사를 찾아 반환합니다."""
    common_friends = {}
    target_interests = set(all_users[target_user])

    for other_user, interests in all_users.items():
        if target_user == other_user:
            continue
        
        common = target_interests.intersection(set(interests))
        if common:
            common_friends[other_user] = list(common)
            
    return common_friends

def find_friends_with_no_common_interests(target_user, all_users):
    """공통 관심사가 하나도 없는 친구 목록을 찾아 반환합니다."""
    no_common_friends = []
    target_interests = set(all_users[target_user])

    for other_user, interests in all_users.items():
        if target_user == other_user:
            continue
            
        if not target_interests.intersection(set(interests)):
            no_common_friends.append(other_user)
            
    return no_common_friends

target = "Alice"

# 1. 공통 관심사를 갖는 친구 응답
common_list = find_friends_with_common_interests(target, user_interests)
print(f"'{target}'님과 공통 관심사를 가진 친구:")
for friend, interests in common_list.items():
    print(f"  - {friend} (공통 관심사: {', '.join(interests)})")

print("-" * 30)

# 2. 공통 관심사가 없는 친구 응답
no_common_list = find_friends_with_no_common_interests(target, user_interests)
print(f"'{target}'님과 공통 관심사가 없는 친구:")
print(f"  - {', '.join(no_common_list)}")
````

### 반복문(Iteration statement)
* 여러 개의 숫자를 입력받아 합계를 계산하는 함수를 작성
* 사용자가 'q'를 입력하면 입력을 중단하고 지금까지 입력한 숫자의 합을 출력

````python
def sum_until_q():
    """
    사용자가 'q'를 입력할 때까지 숫자를 입력받아 합계를 계산하고 출력합니다.
    """
    total = 0
    
    while True:
        user_input = input("숫자를 입력하세요 (종료하려면 'q' 입력): ")
        
        if user_input.lower() == 'q':
            print("입력을 종료합니다.")
            break

        try:
            number = float(user_input)
            total += number
        except ValueError:
            print("잘못된 입력입니다. 숫자 또는 'q'만 입력해주세요.")
    
    print(f"\n지금까지 입력한 숫자의 총합은 {total} 입니다.")

sum_until_q()
````
