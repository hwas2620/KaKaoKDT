# FAST API
## REST API 기초
### REST란?
* REST(Representational State Transfer)는 웹 서비스를 설계하는 아키텍처 스타일
    * 자원(Resource)의 표현(Representation)에 의한 상태(State) 전달(Transfer)을 의미
* 식당 주문 시스템과 유사
    * 메뉴판(API 문서)에서 원하는 음식(리소스)을 선택
    * 주문서(요청)를 작성해 전달
    * 음식(응답)이 제공
* REST의 탄생 배경
    * 2000년 Roy Fielding의 박사 논문에서 처음 제시됨
    * 웹의 장점을 최대한 활용할 수 있는 아키텍처로 설계
    * 분산 시스템 간의 효율적인 통신을 위해 개발
### REST의 핵심 개념
* 표현(Representation)
    * 실제 자원은 서버에 저장되어 있지만, 클라이언트는 자원의 '표현'을 주고받음
    * 같은 자원이라도 다양한 형식(JSON, XML 등)으로 표현
    * 같은 상품 정보가 웹 브라우저에선 HTML로, 모바일 앱에선 JSON으로 표현
* 상태 전이(State Transfer)
    * REST에서 '상태 전이'란 클라이언트가 현재 상태에서 다음 상태로 어떻게 전환하는지를 의미
    * 사용자는 하이퍼링크를 통해 다양한 리소스 간 이동하며 애플리케이션 상태를 변경
        * 서버는 응답에 다음에 사용 가능한 상태(링크)를 포함
        * 클라이언트는 이 링크를 통해 다음 상태로 전이합니다
        * 이 방식으로 클라이언트는 미리 모든 API 주소를 알 필요 없이 응답에 포함된 링크를 따라갈 수 있습니다
    * 예시 : 쇼핑몰에서 사용자가 상품 목록 → 상품 상세→ 장바구니 → 결제 페이지로 이동하는 과정
* 무상태성(Stateless)
    * 각 요청은 독립적이며, 서버는 클라이언트의 상태를 저장하지 않음
    * 예시 : 로그인 후 API 호출 시 매번 인증 토큰을 함께 보내야 함
* 통합 인터페이스(Uniform Interface)
    * REST는 리소스를 조작하는 일관된 방법을 제공
    * HTTP 메서드(GET, POST, PUT, DELETE)를 사용해 모든 리소스를 같은 방식으로 처리
    * 모든 가전제품의 전원 버튼이 비슷한 위치에 있는 것처럼, REST API는 모든 리소스에 대해 일관된 인터페이스를 제공
### API란
* API(Application Programming Interface)는 서로 다른 프로그램이 상호작용하는 방법을 정의한 규약
    * 표준화된 인터페이스를 제공하여 다양한 기기를 연결할 수 있게 합니다.
    * 전기 콘센트에 연결하듯, 내부 작동 방식(전기가 어떻게 생성되는지)을 알 필요 없이 사용할 수 있습니다.
### REST API 설계 원칙
* 클라이언트-서버 구조
    * 관심사의 분리 : UI와 데이터 저장은 분리되어야 함
    * 독립적인 진화 : 클라이언트와 서버는 독립적으로 발전 가능
* 무상태(Stateless)
    * 각 요청은 완전한 정보를 담고 있어야 함
    * 서버는 세션 정보를 저장하지 않음
    * 서버 확장성이 높아짐
* 캐시 가능성(Cacheable)
    * 응답은 캐싱 가능 여부를 명시해야 함
    * 효율적인 캐싱은 성능 향상에 중요
* 계층화 시스템(Layered System)
    * 클라이언트는 서버와 직접 연결되었는지, 중간 서버와 연결되었는지 알 수 없음
    * 중간 계층(로드 밸런서, 캐시 등)을 추가하여 확장성과 보안 강화 가능
* 일관된 인터페이스(Uniform Interface)
    * 리소스 식별
    * HATEOAS(Hypermedia as the Engine of Application State)
        * 다음에 할 수 있는 링크를 같이 제공
### HTTP 메서드
* 도서관 책 관리 API 구축 시,
    * GET: 책 찾아보기 / 리소스 조회
    * POST: 새 책 등록하기 / 리소스 생성
    * PUT: 책 정보 전체 수정하기 / 리소스 전체 수정
    * PATCH: 책 정보 일부 수정하기 / 리소스 일부 수정
    * DELETE: 책 폐기하기 / 리소스 삭제
### HTTP 상태 코드
* 상태 코드는 요청의 성공, 실패 등의 상태를 나타냄

### URI 설계 원칙
* URI(Uniform Resource Identifier) : 리소스를 식별하는 고유한 주소
* 명사형 리소스 사용**
    * `/users` (O), `'/getUsers` (X)
    * `'/products` (O), `/createProduct` (X)
* 복수형 명사 사용
    * `/users` (O), `/user` (X)
    * `/products` (O), `/product` (X)
* 계층 관계 표현
    * `/users/123/orders` (특정 사용자의 주문 목록)
    * `/departments/HR/employees` (HR 부서의 직원 목록)
* 소문자 사용
    * `/products` (O), `/Products` (X)
* 하이픈(-) 사용, 언더스코어(_) 지양**
    * `/product-categories` (O), `/product_categories` (X)
* 파일 확장자 포함하지 않기
    * `/users` (O), `/users.json`` (X)
### 예시: 온라인 쇼핑몰 API 설계
* 사용자 목록
    * `/users` : 모든 사용자 정보
* 특정 사용자
    * `/users/123`: ID가 123인 사용자 정보
* 사용자의 주문
    * `/users/123/orders`: ID가 123인 사용자의 주문 목록
* 특정 주문
    * `/orders/456`: ID가 456인 주문 정보
* 카테고리별 제품
    * `/product-categories/electronics/products` : 전자제품 카테고리의 제품 목록
### RESTful과 non-RESTful API
* RESTful API
    * GET /users/123
* 리소스 중심 설계 : /users/123은 명사형으로 특정 자원(사용자)을 식별
* URI 계층 구조 : 명확한 계층 구조로 리소스를 표현
* HTTP 메서드 적절한 사용: GET 메서드를 조회 목적으로 올바르게 사용합니다.
* non-RESTful API
    * GET /getUser?id=123
* 동사 사용 : getUser라는 동사를 사용하여 행위 중심으로 설계
* HTTP 메서드와 중복 : get이라는 동사를 URI에 포함
* 쿼리 파라미터 사용: 리소스 식별을 위해 경로 변수 대신 쿼리 파라미터를 사용
## FAST API 소개 및 설치
### FastAPI란
* FastAPI는 Python으로 웹 API를 빠르고 쉽게 만들 수 있는 현대적인 웹 프레임워크
### FastAPI의 주요 특징
* 빠른 성능
    * Node.js와 Go 수준의 높은 성능
* 자동 문서화
    * Swagger UI를 통한 대화형 문서 제공
* 타입 안정성
    * 런타임에서 타입 체크

## FastAPI API 문서화

#### 과제
* 카페 주문 관리 시스템의 API를 설계하고 문서화
    * 음료 메뉴 조회 기능
    * 새 주문 생성, 주문 조회 기능
    * 기능별로 그룹화
    * 각 API의 용도와 사용법 설명
    * 실제 사용 가능한 예시 포함
* 구현해야 할 엔드포인트**:
    * `GET /menu` - 전체 메뉴 조회
    * `GET /menu/{item_id}` - 특정 메뉴 아이템 조회
    * `POST /orders` - 새 주문 생성
    * `GET /orders/{order_id}` - 주문 상세 조회
