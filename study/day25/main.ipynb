### 인스턴스 세그멘테이션
#### YOLO
### 객체 검출
* 이미지에서 관심 있는 객체의 위치와 클래스를 동시에 예측하는 기술
* 사진에서 사람이나 자동차 같은 것들을 찾아서 네모 상자로 표시하고, "이건 사람이야", "저건
자동차야"라고 이름표를 붙이는 작업
* 객체 검출 : 빠름, 가벼움, 바운딩 박스만
* 인스턴스 세그먼테이션 : 느림, 무거움, 정확한 마스크
#### 바운딩 박스
* 바운딩 박스는 객체를 감싸는 최소 크기의 직사각형으로, 보통 (x1, y1, x2, y2) 또는 (x, y, width, height)
형태로 표현
#### IoU (Intersection over Union)
* loU는 두 바운딩 박스의 겹치는 영역과 전체 영역의 비율로, 객체 검출의 정확도를 측정하는 핵심 지표
    * 정답박스와 예측박스가 겹치는 비율로 위치 정보까지 정확도를 측정 가능
* IoU = (교집합 영역) / (합집합 영역)
#### NMS (Non-Maximum Suppression)
* NMS는 객체 검출에서 중복된 바운딩 박스를 제거하는 후처리 기법
* 같은 객체에 대해 여러 개의 검출 결과가 나올 때, 가장 좋은 것만 남기고 나머지는 제거
* 한 사람을 여러 번 동그라미 쳤을 때, 겹치는 동그라미들 중에서 가장 정확한 하나만 남기고 나머지는 지우는 것
* NMS 알고리즘 과정
    * 신뢰도(confidence) 기준으로 바운딩 박스 정렬
    * 가장 높은 신뢰도의 박스 선택
    * 선택된 박스와 IoU가 임계값 이상인 박스들 제거
    * 남은 박스들로 2-3 과정 반복
#### mAP (mean Average Precision)
* 객체 검출 모델의 성능을 종합적으로 평가하는 지표로, 모든 클래스에 대한 Average Precision의 평균
* 즉, 모든 클래스에 대해 평균적으로 얼마나 정확하게 객체를 탐지했는지를 나타내는 지표
* 계산 과정
    * Precision-Recall 곡선 : 각 클래스별로 정밀도와 재현율의 관계를 그래프로 표현
    * Average Precision (AP): PR 곡선 아래 면적
    * mAP: 모든 클래스의 AP 평균
#### Precision-Recall Tradeoff
* 정밀도(Precision)와 재현율(Recall) 정의
    * 정밀도(Precision)
        * 모델이 "양성"이라고 예측한 것 중에서 실제로 양성인 비율
        * TP / (TP + FP) - "예측의 정확성"
    * 재현율(Recall)
        * 실제 양성인 것 중에서 모델이 올바르게 찾아낸 비율
        * TP / (TP + FN) - "실제 양성의 발견율"
* TradeOff 발생 이유
    * 임계값을 낮추면 (더 관대하게 예측):
        * 더 많은 샘플을 "양성"으로 예측 → 재현율↑
        * 하지만 잘못된 예측(False Positive)도 증가→ 정밀도↓
    * 임계값을 높이면 (더 엄격하게 예측):
        * 확실한 것만 "양성"으로 예측 → 정밀도 ↑
        * 하지만 실제 양성을 놓치는 경우(False Negative) 증가→ 재현율↓
