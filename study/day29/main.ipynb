## 관계형 데이터베이스
### 데이터베이스
* 체계적으로 구조화되고 관리되는 데이터의 집합
### 관계형 데이터베이스 (RDBMS)
* 테이블 형태로 데이터를 저장하고, 테이블 간의 관계를 통해 데이터를 관리하는 시스템
* 테이블 : 행(row)과 열(column)로 구성된 2차원 표 형태
* 행(Row/Record) : 하나의 완전한 데이터 항목 (예: 한 명의 사용자 정보)
* 열(Column/Field) : 데이터의 속성 (예: 이름, 나이, 이메일)
* 기본 키(Primary Key) : 각 행을 고유하게 식별하는 값
* 외래 키(Foreign Key) : 다른 테이블과의 관계를 나타내는 키
### SQL이란?
* Structured Query Language의 약자로 구조적 질의 언어를 의미
* 즉, 관계형 데이터 베이스(RDB)에 질의를 하기 위한 언어
### 릴레이션
* 릴레이션이란 관계형 모델에서 데이터를 표현하는 방식
    * 제목(Head)과 본체(Body)로 구성
    * 제목(Head)은 속성(attribute)이 n개 모여있는 집합이며, 속성은 해당 속성명과 데이터 타입 2가지로 구성
    * 본체(Body)는 속성 값의 집합이며, 본체의 속성 값 하나하나를 튜플(Tuple)이라고 함
    * 즉, 릴레이션이란 제목과 본체의 쌍을 가진 관계입니다.
### SQL과 관계형 모델의 비교
* 아래 그림에서 확인 가능한 SQL 테이블이 바로 관계형 모델에서의 릴레이션
* 이름/나이/성별로 구분되어 있는 열(Column)이 속성
* 해당 속성의 값들인 각각의 행(Row)들이 튜플 ex) 다한/25/남자
* 도메인이란 속성들이 가질 수 있는 값들의 범위가 지정되어 있는 집합
    * 예를 들어 성별이라는 속성이 있으면 해당 속성이 가질 수 있는 값은 남자 또는 여자뿐
### 주요 명령
* SELECT
    * SQL에서 데이터를 조회하는데 사용하는 유일한 명령어
* INSERT
    * SQL 테이블의 하나의 행을 추가하는 것
* DELETE
    * SQL 테이블의 하나의 행을 삭제하는 것
* UPDATE
    * SQL 테이블의 특정 튜플을 수정하는 것
### 데이터베이스 정규화
* 정규화 이론이 가지고 있는 특정한 규칙을 바탕으로 관계형 모델을 구성하는 것
* 제1 정규형(1NF)
* 제2 정규형(2NF)
* 제3 정규형(3NF)
* 보이스코드 정규형(BCNF)
* 제4 정규형(4NF)
* 제5 정규형(5NF)
* 제6 정규형(6NF)
### 데이터베이스 정규화
* 아래 릴레이션을 보면 정은이 2학년과 3학년에 동시에 속해있음
* 따라서 해당 릴레이션을 통해 정은이가 2학년인지 3학년인지 답을 할 수 없음
* 데이터의 중복을 최대한 피하면서 DB를 설계해야 하고, 이렇게 데이터의 중복을 제거하는 데에 도움이 되는 것이 바로 정규화 이론

### 제1 정규형(1NF)
* 행이 위에서 아래로 정렬돼 있지 않아야 한다
* 열이 왼쪽에서 오른쪽으로 정렬돼 있지 않아야 한다
    * 실제로 SQL에서는 행과 열에 순서가 존재
    * 하지만 여기서 말하는 "정렬돼 있지 않다"의 의미는 테이블 순서에 상관없이 연산되어야 한다는 것 따라서 쿼리문을 작성할 때 행과 열의 위치에 의존하지 않도록해야 함
* 중복하는 행이 존재하지 않아야 한다
* 각 행과 열의 교차점(열의 값)은 도메인(데이터형)에 속하는 요소의 값을 딱 한 개만 가진다
    * 이론적으로 요소의 값은 값이 원자성을 가져야 합니다.
    * 즉 더 이상 분해할 수 없는 최소 단위여야 함
* 모든 열의 값은 정의된 것이어야 하고 각 행은 항상 존재 한다
## 데이터베이스 연동
### MySQL
* 세계에서 가장 널리 사용되는 오픈 소스 관계형 데이터베이스 관리 시스템
* 무료 오픈 소스 : 라이선스 비용 없이 사용 가능
* 높은 성능 : 빠른 데이터 처리 속도
* 안정성 : 대용량 데이터 처리에 적합
* 크로스 플랫폼 : Windows, Linux, macOS 등 다양한 운영체제 지원
* 표준 SQL 지원 : 표준 SQL 문법 사용 가능
### ORM이란?
* ORM(Object-Relational Mapping)은 객체 지향 프로그래밍 언어와 관계형 데이터베이스 간의 데이터를
* 변환하는 기술
* Query 직접 작성
    * 복잡한 SQL 문 직접 작성 필요
    * "SELECT * FROM users WHERE age > 25 AND name LIKE '%김%'"
* ORM 사용
    * 간단한 Python 코드로 작성
    * users = session.query(User).filter(User.age > 25, User.name.contains('김'))
### SQLAlchemy
* Python에서 가장 인기 있는 ORM 라이브러리
* 생산성 향상 : SQL 문을 직접 작성할 필요 없음
* 타입 안전성 : Python 타입 시스템 활용
* 데이터베이스 독립성 : MySQL, PostgreSQL, SQLite 등 다양한 DB 지원
* 객체 지향적 : Python 클래스로 테이블 정의
* 관계 관리 : 테이블 간 관계를 쉽게 정의하고 사용
* 설치
    * pip install sqlalchemy
    * pip install pymysql
