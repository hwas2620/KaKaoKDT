## 요청 및 응답 처리
### 파일 업로드
* 클라이언트가 서버로 파일을 전송하는 기능
    * 이미지, 문서, 동영상 등 다양한 파일 업로드 가능
    * 파일 크기, 형식 제한 설정 가능
    * 업로드된 파일을 서버에 저장하거나 처리
* 파일 업로드를 위한 모듈 설치
    * pip install python-multipart
* UploadFile 클래스 활용
    * FastAPI에서 제공하는 파일 업로드 전용 클래스
    * 파일명, 크기, 타입 등의 메타데이터와 실제 파일 내용을 포함
* UploadFile의 주요 속성
    * `filename`: 업로드된 파일의 원본 이름
    * `content_type`: 파일의 MIME 타입 (예: image/jpeg, text/plain)
        * MIME(Multipurpose Internet Mail Extensions) 타입 : 파일의 형식과 성격을 나타내는 표준 식별자
    * `size`: 파일 크기 (바이트 단위)
* UploadFile의 주요 메서드
    * `read()`: 파일 전체 내용을 바이트로 읽기
    * `seek(0)`: 파일 포인터를 처음으로 이동
    * `close()`: 파일 스트림 닫기

## 의존성 주입 시스템
### 의존성 주입(Dependency Injection)이란?
* 의존성 주입의 장점
    * 코드 재사용성 : 공통 로직을 여러 엔드포인트에서 재사용
    * 유지보수성 : 변경사항이 한 곳에서만 발생
    * 테스트 용이성 : Mock 객체로 쉽게 테스트 가능
    * 관심사 분리 : 비즈니스 로직과 부가 기능 분리
### Depends 클래스
* FastAPI에서 의존성 주입을 처리하는 핵심 클래스
* Depends 클래스의 특징
    * 자동 호출 : 의존성 함수를 자동으로 호출
    * 캐싱 : 같은 요청 내에서 동일한 의존성은 한 번만 실행
    * 타입 힌트 지원 : IDE에서 자동완성과 타입 검사 제공
    * 중첩 지원 : 의존성이 다른 의존성을 가질 수 있음
### Depends 클래스의 캐싱
* 캐싱 : 한 번 계산한 결과를 임시로 저장해두고, 같은 계산이 필요할 때 다시 계산하지 않고 저장된 결과를 재사용하는 기법
* FastAPI 캐싱의 특징
    * 요청 범위 캐싱 : 하나의 HTTP 요청 내에서만 캐시가 유지됨
    * 자동 관리 : 개발자가 별도로 캐시를 관리할 필요 없음
    * 메모리 효율성 : 요청이 끝나면 캐시도 자동으로 정리됨
요청 내 캐싱 (Request-scoped Caching)
    /test2 엔드포인트에서 같은 의존성을 두 번 사용하지만 함수는 한 번만 실행됨
    data1과 data2는 완전히 동일한 객체 (같은 메모리 주소)
요청 간 캐싱 없음 (No Cross-request Caching)
    /test1을 여러 번 호출하면 매번 새로 함수가 실행됨
    각 HTTP 요청은 독립적인 캐시를 가짐
캐싱 범위의 한계
    캐시는 하나의 HTTP 요청이 처리되는 동안만 유지됨
    요청이 끝나면 캐시도 함께 사라짐
캐시 비활성화
    시간에 따라 변하는 데이터 : 현재 시간, 랜덤값등
    상태가 변하는 객체 : 카운터, 세션 정보 등
    부작용이 있는 함수 : 로그 기록, 외부 API 호출 등
### Depends와 타입 힌트
타입 힌트는 함수의 매개변수와 반환값의 데이터 타입을 명시하는 Python 기능
FastAPI에서 타입 힌트가 중요한 이유
    자동 문서 생성 : Swagger/OpenAPI 문서에 정확한 타입 정보 표시
    데이터 검증 : 잘못된 타입의 데이터가 들어오면 자동으로 에러 발생
    코드 가독성 : 다른 개발자가 코드를 이해하기 쉬움
    런타임 안정성 : 타입 관련 오류를 사전에 방지

        관리자만 접근 가능한 기능
