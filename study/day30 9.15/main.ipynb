과제 1

과제 2
기존 VM에서 새로운 VM으로 연결
  - ping 명령어를 입력하여 연결을 확인 합니다.
    ping 192.168.56.210

  - ssh 명령어를 이용하여 기존 VM 에서 새 VM으로 연결 합니다.
    ssh root@192.168.56.210 

  - 새로운 VM의 호스트 명을 ec2 로 설정합니다.
    hostnamectl set-hostname ec2 
  
  - bash 를 새롭게 실행해서 호스트 명이 변경되는 것을확인합니다.
    bash

  - 작업 완료후 빠져 나와서 다시 연결합니다.
    exit 
    ssh root@192.168.56.210

추가 ssh key 기반 인증
기존 vm에서 실행합니다. 
ssh-keygen -N '' -f ~/.ssh/id_rsa
ssh-copy-id root@192.168.56.210
