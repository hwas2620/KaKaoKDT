```python
### pandas
import pandas as pd
print(pd.__version__)
```

```python
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])

# 기본 속성
print("값 배열:", s.values) # [10 20 30 40 50]
print("인덱스:", s.index) # Index(['a', 'b', 'c', 'd', 'e'], dtype='object')

# 기본 통계 메소드
print("평균:", s.mean())  # 30.0
print("합계:", s.sum())   # 150
print("최소값:", s.min()) # 10
print("최대값:", s.max()) #50

# 데이터 접근
print("'c' 인덱스의 값:", s['c']) #30
print("여러 인덱스의 값:", s[['a', 'c', 'e']]) #a 10, c 30, e 50
```

```python
### dataframe
import pandas as pd
print(pd.__version__)

data = {
    ' Name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'Age': [28, 24, 35, 32, 45],
    'City': ['New York', 'Paris', 'Berlin', 'London', 'Tokyo'],
    'Salary': [50000, 65000, 75000, 85000, 60000]
}
df = pd.DataFrame(data)
print(df)
```

```python
# 샘플 DataFrame 생성 - 직원 정보를 담은 테이블 형태의 데이터
df = pd.DataFrame({
    'Name': ['John', 'Anna', 'Peter', 'Linda','Bob'],
    'Age':[28, 24,35,32, 43],
    'City': ['New York', 'Paris','Berlin','London','Tokyo'],
    'Salary':[50000,65000,75000,85000,60000],
    'Department':['IT','HR', 'IT','Finance', 'Marketing']
})
dept_groups = df.groupby("Department")
print(dept_groups)
print(dept_groups['Salary'].transform("mean"))
```
