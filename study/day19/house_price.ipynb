```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
#sklearn: 머신러닝 라이브러리
# pip install scikit-learn

#간단한 집값 예측 머신러닝 모델
def house_price_prediction():
    """
    집 크기를 기반으로 집값을 예측하는 머신러닝 모델
    """

    # 가상의 집 데이터 생성 (크기 -> 가격)
    np.random.seed(42)
    #랜덤 시드는 랜덤 함수의 초기 상태를 지정하는 값
    # 동일한 랜덤 시드를 사용하면 항상 동일한 난수 시퀀스를 생성
    #42 : 은하수를 여행하는 히치하이커를 위한 안내서에서 '삶, 우주, 그리고 모든 것에 대한 궁극적인 질문의 해답'
    # house_sizes : 집 크기
    house_sizes = np.random.normal(100, 30, 1000) # 평균 100평, 표준편차 30, 1000개의 데이터
    print(house_sizes)
    # house_prices: 집 가격
    #크기가 클수록 가격이 높아지는 관계 + 노이즈
    #노이즈 : 데이터에 추가된 무작위 오차
    # 실제로는 집값이 크기에만 의존하지 않기 때문에 현재는 노이즈를 추가하는 형태로 데이터를 생성
    house_prices = house_sizes * 50 + np.random.normal(0, 500, 1000) + 2000

    print(house_prices)

    # 데이터 전처리
    x = house_sizes.reshape(-1, 1) # 2D 배열로 변환 (행, 열)
    # -1 : 이 차원의 크기를 자동으로 계산하라는 의미
    #1: 열의 개수를 1개로 고정
    # 원본 배열의 모든 원소를 1개 열로 재배치하여 2D 배열 생성
    # scikit-learn의 모델들은 2D 배열을 입력으로 받아야 함
    # (n_samples, n_features) 형태가 필요
    # 여기서는 (1000개 샘플, 1개 특성)으로 변환
    y = house_prices

    #훈련용/테스트용 데이터 분할
    x_train, x_test, y_train, y_test = train_test_split(
        x, y, test_size=0.2, random_state=42
    )
    # train_test_split 함수 매개변수
    #x: 훈련 데이터
    #y : 훈련 데이터의 정답
    # test_size: 테스트 데이터 비율
    # random_state : 데이터 분할 시 랜덤 시드 설정

    # 머신러닝 모델 생성 및 훈련
    model = LinearRegression()
    # LinearRegression 클래스의 fit 메서드
    #모델 훈련
    #x: 훈련 데이터
    #y : 훈련 데이터의 정답
    model.fit(x_train, y_train)
    # 예측
    y_pred = model.predict(x_test)

    #성능 평가
    mse = mean_squared_error(y_test, y_pred)
    #MSE: 실제값과 예측값의 차이를 제곱한 후 평균을 계산
    #모델이 예측을 잘하는지 평가하는 지표
    #값이 작을수록 예측 성능이 좋음
    r2 = r2_score(y_test, y_pred)
    #R: 모델이 데이터의 변동을 얼마나 설명하는지 평가하는 지표 (집크기에 따른 예측 값이 잘 달라지는 것을 의미)
    #1에 가까울수록 모델이 데이터의 변동을 잘 설명

    print(f"평균 제곱 오차 (MSE): {mse:.2f}") # .2f: 소수점 2자리까지 출력
    print(f"결정 계수 (R2): {r2:.2f}")
    print(f"모델 계수 (기울기): {model.coef_[0]:.2f}")
    #coef: 입력 특성(집 크기)이 1 단위 증가할 때 예측값이 얼마나 변하는지
    # 예: coef [0] 50.0 이면 "집 크기 1명 증가 집값 50만원 증가"
    print(f"모델 절편: {model.intercept_:.2f}")
    #intercept: 입력 특성이 일 때의 예측값 (기본값)
    #예: intercept 2008.0 이면 "집 크기가 평일 때 기본 가격 2000만원"

    # 새로운 집 크기에 대한 예측
    new_house_sizes = [80, 120, 150]
    for size in new_house_sizes:
        predicted_price = model.predict([[size]])[0]
        # [[size]]
        # 예측을 위해 2D 배열 형태로 전달
        # 모델은 하나의 샘플에 대해 예측하므로 2D 배열 필요
        print(f"{size}평 집의 예상 가격: {predicted_price:.2f}만원")

    return model, x_test, y_test, y_pred

#실행
model, x_test, y_test, y_pred = house_price_prediction()
```
