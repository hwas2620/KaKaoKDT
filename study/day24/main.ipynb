# OCR 심화
## TrOCR
* TrOCR (Transformer-based Optical Character Recognition)
    * Vision Transformer와 Text Transformer를 결합한 최신 End-to-End OCR 모델
    * Microsoft Research (2021년 발표)
    * 이미지 패치를 토큰으로 처리하는 완전한 Transformer 기반 구조
* 완전한 Transformer: CNN/RNN 없이 Transformer만 사용
* 사전 훈련 모델 : 대규모 데이터로 미리 학습된 모델 제공
* End-to-End : 복잡한 파이프라인 없이 이미지 → 텍스트 직접 변환
* 어텐션 내장 : 별도 구현 없이 Self-Attention 메커니즘 활용

### 앙살블 모델
* 여러 개의 서로 다른 모델을 조합하여 더 우수한 성능을 얻는 기계학습 기법
* 단일 OCR 모델로는 모든 상황에서 완벽한 성능을 보장하기 어렵기 때문에, 여러 모델의 장점을 결합한 앙상블 접근법이 필요
* 언어별 특화
    * PaddleOCR → 한글에 강함
    * EasyOCR → 영어에 강함
    * TrOCR → 고품질 인쇄체에 강함
* 문서 유형별 특화
    * 명함: PaddleOCR + 후처리
    * 스크린샷: EasyOCR + 전처리
    * 스캔 문서: TrOCR + 품질 향상
* 상황별 보완
    * 조명이 나쁜 경우 : 전처리 + 여러 모델
    * 기울어진 텍스트 : PaddleOCR 우선 + EasyOCR 보조
    * 혼용 텍스트 : 모든 모델 결합 후 최적 선택
### 앙상블 모델 방법론
* 투표 방식
    * 더 비슷한(동일한) 결과가 많은 쪽을 선택하는 다수결 방식
    * 장점
        * 간단하고 직관적
        * 명확한 다수결이 있을 때 매우 효과적
        * 개별 모델의 일시적 오류를 보정
    * 단점
        * 모든 결과가 다르면 판단 불가
        * 신뢰도를 고려하지 않음
        * 미묘한 차이(띄어쓰기 등)도 다른 것으로 판단
* 신뢰도 기반 방식
    * 가장 높은 신뢰도를 가진 결과 선택
    * 장점
        * 모델의 확신 정도를 반영한 합리적 선택
        * 모든 결과가 달라도 판단 가능
        * 특정 상황에서 한 모델이 특히 확신할 때 유용
    * 단점
        * 신뢰도 점수가 항상 정확하지 않을 수 있음
        * 과신하는 모델이 있으면 편향된 결과 가능
        * 신뢰도 계산 방식에 따라 결과가 달라질 수 있음
* 조건부 선택 (Conditional Selection)
    * 입력 데이터의 특성을 분석하여 상황에 가장 적합한 모델을 선택
    * 장점
        * 상황별 최적 모델을 똑똑하게 선택
        * 도메인 지식을 활용한 합리적 판단
        * 각 모델의 강점을 최대한 활용
        * 확장성이 좋음 (새로운 조건 추가 용이)
    * 단점
        * 조건 설정이 복잡하고 전문 지식 필요
        * 조건 임계값 설정이 주관적일 수 있음
        * 유지보수가 어려울 수 있음
