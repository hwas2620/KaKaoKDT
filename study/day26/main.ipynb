#### 대화 상태 관리
* 다중 턴 대화에서 사용자의 의도, 이전 발화 내용, 대화 진행 상황을 추적하고 관리하는 기술
* 대화 상태의 구성 요소
    * Intent History: 의도 변화 추적
    * Entity Memory: 추출된 개체 정보 저장
    * Context Stack: 대화 맥락 유지
    * User Profile: 사용자 개인화 정보
### Agent
* 특정 환경에서 자율적으로 작동하며, 목표를 달성하기 위해 인식(Perception), 추론(Reasoning), 행동(Action)을 수행할 수 있는 지능형 개체
* AI 에이전트는 주어진 상황을 분석하고, 최적의 행동을 선택하여 실행하며, 그 결과를 학습하여 성능을 개선
#### 에이전트 기반 문제 해결
* 복잡한 문제를 단계별로 분해하여 계획을 수립하고, 각 단계를 체계적으로 실행하며, 결과를 평가하는 자율적인 AI 시스템
* 일반 챗봇과 달리, 에이전트는 여러 단계에 걸쳐 복잡한 작업을 체계적으로 처리
#### 핵심 개념
* 계획 수립 (Planning)
    * 목표 분석 및 하위 작업 분해
    * 작업 우선순위 결정
    * 리소스 할당 및 제약 사항 고려
* 실행 (Execution)
    * 계획에 따른 순차적 작업 수행
    * 도구 사용 및 API 호출
    * 중간 결과 저장 및 전달
* 평가 (Evaluation)
    * 작업 결과 검증
    * 성공/실패 판단
    * 개선 방안 도출
#### 일반 챗봇 vs 에이전트 시스템
* 작업 처리 방식
    * 일반 챗봇: 질문 → 즉시 답변 (1회성 반응)
    * 에이전트 시스템: 목표 → 계획 수립 → 단계별 실행 → 평가 (다단계 프로세스)
* 자율성 (Autonomy)
    * 일반 챗봇 : 사용자 입력에만 반응하는 수동적 시스템
    * 에이전트 시스템 : 목표 달성을 위해 스스로 판단하고 행동하는 능동적 시스템
* 환경과의 상호작용
    * 일반 챗봇 : 텍스트 입출력에 국한
    * 에이전트 시스템 : 외부 도구, API, 데이터베이스 등과 실제 상호작용
* 메모리와 학습
    * 일반 챗봇 : 대화 히스토리 정도의 단순한 기억
    * 에이전트 시스템 : 경험을 축적하고 학습하여 성능 개선
* 여행 계획을 세워줘"라는 요청에 대해
    * 일반 챗봇
        * "여행 계획을 위해서는 목적지, 예산, 일정을 고려해야 합니다. 일반적으로 항공편, 숙박, 관광지를 미리 예약하는 것이 좋습니다."
    * 에이전트 시스템
        * 계획 수립 : 사용자 선호도 파악, 예산 분석, 일정 확인
        * 실행
            * 실시간 항공료 검색 API 호출, 숙박 예약 사이트에서 가격 비교, 날씨 정보 확인, 현지 이벤트 조사
        * 평가 : 옵션들을 비교분석하여 최적안 도출
        * 결과 : 구체적인 여행 일정표와 예약 링크 제공
